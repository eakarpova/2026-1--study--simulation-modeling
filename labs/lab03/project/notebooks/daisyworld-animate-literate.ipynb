# Анимация модели Daisyworld

**Цель:** Создать видео эволюции модели, показывающее динамику изменения температуры и распределения маргариток.

## Инициализация
Модель создаётся с параметрами по умолчанию.

## Результаты
Анимация сохранена в файл:
[simulation.mp4](../project/plots/simulation.mp4)

## Вывод
Видео демонстрирует, как чёрные и белые маргаритки регулируют температуру планеты, поддерживая её в пригодном для жизни диапазоне.
