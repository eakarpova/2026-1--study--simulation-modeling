# Динамика модели при изменении светимости

**Цель:** Исследовать, как изменение солнечной светимости влияет на численность маргариток и температуру.

## Инициализация
Модель запускается со сценарием `:ramp` (изменение светимости).

## Результаты
![](../project/plots/daisy_luminosity.png)

## Вывод
При увеличении светимости чёрные маргаритки замещаются белыми, компенсируя нагрев. Система саморегулируется.
