# Динамика численности маргариток

**Цель:** Построить график изменения численности чёрных и белых маргариток во времени.

## Инициализация
Модель запускается на 1000 шагов.

## Результаты
![](../project/plots/daisy_count.png)

## Вывод
Популяции колеблются, но в среднем поддерживается равновесие между чёрными и белыми маргаритками.
